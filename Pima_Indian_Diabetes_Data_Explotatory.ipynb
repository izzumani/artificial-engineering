```python
import numpy as np # linear algebra
import pandas as pd # data processing, CSV file I/O (e.g. pd.read_csv)
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.feature_selection import VarianceThreshold
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
data = pd.read_csv('/content/sample_data/Pima_Indian_Diabetes.csv')
data.head()
```

```python
data.describe()
```

```python
data.dtypes
```

```python
data.isna().sum()
```

```python
data = data.rename(columns={'Pregnancies': 'pregnant', 'Glucose': 'glucose', 'BloodPressure': 'bp', 'SkinThickness': 'skin', 'Insulin': 'insulin', 'BMI': 'bmi', 'DiabetesPedigreeFunction': 'pedigree', 'Age': 'age', 'Outcome': 'label'})
```

```python
X = data.drop(labels=['label'], axis=1)
y = data['label']
X.head()
```

```python
plt.figure(figsize=(10,10))
sns.barplot(x=y, y='age', data=X)
```

```python
plt.figure(figsize=(10,10))
sns.countplot(x=y, data=X)
```

```python
plt.figure(figsize=(10,10))
sns.boxplot(x=y, y='glucose', data=X)
#
```
