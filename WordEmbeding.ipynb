```python
import gensim.downloader as api
from gensim.models import Word2Vec
dataset = api.load("text8")
model = Word2Vec(dataset)
model.save("/content/sample_data/text8-word2vec.bin")
```

Let us reload the Word2Vec model we just built and
 explore it using the Gensim API. The actual word
vectors can be accessed as a custom Gensim class from the model’s wv attribute:

```python
from gensim.models import KeyedVectors
model = KeyedVectors.load("/content/sample_data/text8-word2vec.bin")
word_vectors = model.wv
```

```python
words = word_vectors.key_to_index
print([x for i, x in enumerate(words) if i< 10])
```

```python
def print_most_similar(word_conf_pairs, k):
  for i, (word,conf) in enumerate(word_conf_pairs):
    print(f"word{i+1}:{word},similarity:{conf}")
    print("{:.3f} {:s}".format(conf,word))
    if i >= k-1:
      break

  if k < len(word_conf_pairs):
    print("...")


print_most_similar(word_vectors.most_similar("king"),5)
```

```python
print_most_similar(word_vectors.most_similar(positive=["france","budapest"],negative=["paris"]),1)
```

```python
print(word_vectors.doesnt_match(["hindus","parsis","singapore","christian"]))
```

```python
for word in ["woman","dog","whale","tree"]:
  print("similarity({:s}, {:s}) = {:.3f}".format("man",word,word_vectors.similarity("man", word)))
```

```python
print(print_most_similar(word_vectors.similar_by_word("singapore"),5))
```

```python
print("distance(singapore, malaysia) = {:.3f}".format(word_vectors.distance("singapore","malaysia")))
```
