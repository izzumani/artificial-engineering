```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
url = "https://www.gutenberg.org/cache/epub/2701/pg2701-images.html"
# html_text = requests.get(url).text
html = requests.get(url)
```

```python
soup = BeautifulSoup(html.text, 'html.parser')
title = soup.title.title   # get the title of the page
title
```

```python
html_tags = set([tag.name for tag in soup.find_all()])
html_tags
```

```python
for all_links in soup.find_all('a'):
    for link in all_links:
        print(link.get_text())  
        # print(link.get('href'))
    
```

```python
for b_value in soup.find_all('b'):
    print(b_value.get_text())
```

```python
for li_value in soup.find_all('li'):
    print(li_value.get_text())
```

```python
article = soup.get_text()
```

```python
from nltk.tokenize import word_tokenize
tokens = word_tokenize(article)
lower_tokens = [t.lower() for t in tokens] 
lower_tokens
```

```python
bow_simple = Counter(lower_tokens)
bow_simple
```

```python
from nltk.corpus import stopwords
lower_tokens = [ w for w in lower_tokens if w.isalpha()]
no_stops = [t for t in lower_tokens if t not in stopwords.words('english')]
Counter(no_stops).most_common(10)
```

```python
from nltk.tag import pos_tag
pos_tags = pos_tag(no_stops)
pos_tags[:10]
```

```python
from nltk.stem import WordNetLemmatizer
# Instantiate the WordNetLemmatizer
wordnet_lemmatizer = WordNetLemmatizer()
# Lemmatize all tokens into a new list: lemmatized
Lemmatized = [wordnet_lemmatizer.lemmatize(t) for t in no_stops]
# Create the bag-og-words: bow
bow = Counter(Lemmatized)
# Print the 100 most common tokens
bow.most_common(100)
```

```python
from gensim.corpora.dictionary import Dictionary    
# Create a Dictionary from the articles: dictionary 
dictionary = Dictionary([Lemmatized])
# Select the id for "boat": boat_id
dict_id = dictionary.token2id.get("whale")
print(dict_id)
# Use boat_id with the dictionary to print the word
dictionary.get(dict_id)
```

```python
corpus =[dictionary.doc2bow(_article) for _article in [Lemmatized]]
corpus[:2]
```

```python
bow_doc = sorted(corpus,key=lambda w: w[1],reverse=True)
for bow in bow_doc:
   for word_id, word_count in bow:
       print(dictionary.get(word_id), word_count)
#    print(word_id[1])
```

```python
# Create the defaultdict: total_word_count
from collections import defaultdict
import itertools

total_word_count = defaultdict(int)

for word_id, word_count in itertools.chain.from_iterable(corpus):
    total_word_count[word_id] += word_count

total_word_count
```
