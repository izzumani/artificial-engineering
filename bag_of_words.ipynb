```python
import nltk
import numpy as np
from nltk.corpus import brown
from sklearn.feature_extraction.text import CountVectorizer
```

```python
def splitter(text, num_words):
	words = text.split()
	for i in range(0, len(words), num_words):
		yield ' '.join(words[i:i + num_words])
```

```python

if __name__ == '__main__':
    # Read the data from the Brown corpus
    data = ' '.join(brown.words()[:10000])
    num_words = 2000
    chunks = []
    counter = 0
    text_chunks = splitter(data, num_words)
    for text in text_chunks:
        chunk = {'index': counter, 'text': text}
        chunks.append(chunk)
        counter += 1
    vectorizer = CountVectorizer(min_df=1, max_df=10)
    doc_term_matrix = vectorizer.fit_transform([chunk['text'] for chunk in chunks])
    vocab = np.array(vectorizer.get_feature_names_out())
    print("Vocabulary:")
    print(vocab)
    print("Document term matrix:")
    chunk_names = [f'Chunk-{i}' for i in range(len(chunks))]
    formatted_row = '{:>12}' * (len(chunk_names) + 1)
    print('\n', formatted_row.format('Word', *chunk_names), '\n')

    for word, item in zip(vocab, doc_term_matrix.T):
        # ‘item’ is a ‘csr_matrix’ data structure
        output = [x for x in item.data]
        output = output + [''] * (len(chunk_names) - len(output))  # Ensure output length matches chunk_names length
        print(formatted_row.format(word, *output))
```
