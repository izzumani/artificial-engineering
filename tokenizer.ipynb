```python
import nltk
nltk.download('punkt')
```

```python
text = "Are you curious about tokenization? Let's see how it works! We need to analyze a couple of sentences with punctuations to see it in action."
```

Let’s start with sentence tokenization. NLTK provides a sentence tokenizer, so
let’s import that:

```python
from nltk.tokenize import sent_tokenize
```

```python
sent_tokenize_list = sent_tokenize(text)
```

```python
print(f"Sentence tokenizer:\n", sent_tokenize_list)
      
```

Word tokenization is very commonly used in NLP. NLTK comes with a couple
of different word tokenizers. Let’s start with the basic word tokenizer:

```python
from nltk.tokenize import word_tokenize
print(f"Word tokenizer:\n", word_tokenize(text))
```

If you want to split this punctuation into separate tokens, then you will need
to use the WordPunct tokenizer:

```python
from nltk.tokenize import WordPunctTokenizer
word_punct_tokenizer = WordPunctTokenizer()
print(f"Word punct tokenizer:\n", word_punct_tokenizer.tokenize(text))
```
